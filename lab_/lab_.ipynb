## Load_data

```python
from pathlib import Path
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
data_path = Path("..") / '01_Data' / 'heart_cleaned.csv'
data = pd.read_csv(data_path)
data.head(7)
```

```python
onehot_enc = OneHotEncoder(sparse_output=False)

encoded = onehot_enc.fit_transform(data[['age_group']])

encoded_df = pd.DataFrame(
    encoded,
    columns=onehot_enc.get_feature_names_out(['age_group']),
    index=data.index
)

data = pd.concat(
    [data.drop(columns=['age_group']), encoded_df],
    axis=1
)
data.head(7)
```

```python
std_scaler = StandardScaler()
features = ['age','trestbps','chol','thalach','oldpeak']
data[features] = std_scaler.fit_transform(data[features])
```
